```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
plt.rcParams['font.family'] = "Inter Variable ss02"
```

```python
data = {
    "TEM": {
        "year": [1932, 1934, 1940, 1958, 1959, 1982, 1991, 1999],
        "resolution_A": [5200, 500, 30, 12, 7, 2.0, 1.5, 0.9],
    },
    "Aberration corrected TEM": {
        "year": [1998, 2003, 2015],
        "resolution_A": [1.4, 1.3, 0.4],
    },
    "STEM": {
        "year": [1938, 1968, 1970, 1974, 1987, 1988, 1989, 1994, 1998],
        "resolution_A": [100, 30, 5.2, 2.5, 3.5, 2.4, 2.1, 1.4, 0.95],
    },
    "Aberration corrected STEM": {
        "year": [1999, 2002, 2004, 2007, 2009],
        "resolution_A": [0.9, 0.82, 0.78, 0.55, 0.38],
    },
}
```

```python
COLORS = ['forestgreen', 'royalblue', 'crimson', 'darkorange']
```

```python
plt.figure(figsize=(4, 4), dpi=300)

plt.yscale('log')
plt.ylim(1e4,1e-1)

for i, (k, d) in enumerate(data.items()):
    plt.plot(d['year'], d['resolution_A'], 'o', color=COLORS[i],
             label=k)

plt.xlabel("Year")
plt.ylabel("Resolution [Å]")
plt.legend(loc='lower right', fancybox=False)

plt.show()
```

```python
plt.figure(figsize=(4, 4), dpi=300)

ax = plt.gca()
ax.set_yscale('log')
ax.set_ylim(1e4, 1e-1)

for i, (k, d) in enumerate(data.items()):
    ax.plot(d['year'], d['resolution_A'], 'o', color=COLORS[i], label=k)

ax.set_xlabel("Year")
ax.set_ylabel("Resolution [Å]")
ax.legend(loc='lower right', fancybox=False)

# Common physical length scales (values in Å)
scales = {
    # "Cell": 1e8,          # 10 μm
    "Bacteria": 1e4,     # 1 μm
    "Virus": 3e2,         # 100 nm
    "Protein": 1e1,       # 1 nm
    "Atom": 1,             # 1 Å
}
ax2 = ax.twinx()
ax2.set_yscale('log')
ax2.set_ylim(ax.get_ylim())
ax2.set_yticks(list(scales.values()))
ax2.set_yticklabels(list(scales.keys()))
ax2.set_ylabel("Approximate length scale")

plt.show()
```
